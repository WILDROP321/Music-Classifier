GET THE FEATURES, AND LOAD it into memory

```python
import os
import numpy as np
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from Features import compile_features
from tqdm import tqdm
import joblib
```

```python
# Load and stack 8 image graphs per song
def load_image_stack(image_folder):
    image_names = [
        "chroma_spectrogram.png",
        "chroma_stripe.png",
        "mel_spectrogram.png",
        "mfcc_heatmap.png",
        "mfcc_surface.png",
        "onset_detection.png",
        "pca_mfcc.png",
        "tsne_mfcc.png"
    ]

    images = []
    for name in image_names:
        path = os.path.join(image_folder, name)
        img = load_img(path, target_size=(224, 224))
        img_array = img_to_array(img) / 255.0
        images.append(img_array)

    stacked = np.concatenate(images, axis=-1)  # Shape: (224, 224, 24)
    return stacked

# Load features, scalar scores, and image stack for each song
def load_full_dataset(song_folder, image_folder):
    feature_list = []
    scalar_list = []
    image_list = []
    file_names = []

    all_files = [f for f in os.listdir(song_folder) if f.endswith(".mp3") or f.endswith(".wav")]
    all_files.sort()  # Sort files to maintain consistent order
    
    for filename in tqdm(all_files, desc="Extracting features", unit="song"):
        if filename.endswith(".mp3") or filename.endswith(".wav"):
            full_path = os.path.join(song_folder, filename)
            print(f"Processing: {filename}")
            try:
                features, tension, energy, _ = compile_features(full_path, viz=False)
                img_stack = load_image_stack(image_folder)

                feature_list.append(features)
                scalar_list.append([tension, energy])
                image_list.append(img_stack)
                file_names.append(filename)

            except Exception as e:
                print(f"Error processing {filename}: {e}")

    return (
        np.array(feature_list),
        np.array(image_list),
        np.array(scalar_list),
        np.array(file_names)
    )
```

```python
# To use in training:
song_dir = "TRAINING_DATA_"
image_dir = "IMAGES"
X_features, X_images, X_scalars, X_files = load_full_dataset(song_dir, image_dir)
```

STEP 3: Define the Model (Multi-Input Autoencoder)

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Dense, Dropout, Flatten, Conv2D, MaxPooling2D,
    Concatenate
)

# --- Feature Input (1D vector) ---
feature_input = Input(shape=(X_features.shape[1],), name="feature_input")
x1 = Dense(512, activation='relu')(feature_input)
x1 = Dropout(0.3)(x1)
x1 = Dense(256, activation='relu')(x1)

# --- Image Input (8 stacked images) ---
image_input = Input(shape=(224, 224, 24), name="image_input")
x2 = Conv2D(32, (3, 3), activation='relu', padding='same')(image_input)
x2 = MaxPooling2D((2, 2))(x2)
x2 = Conv2D(64, (3, 3), activation='relu', padding='same')(x2)
x2 = MaxPooling2D((2, 2))(x2)
x2 = Flatten()(x2)
x2 = Dense(256, activation='relu')(x2)

# --- Scalar Input (tension + energy) ---
scalar_input = Input(shape=(2,), name="scalar_input")
x3 = Dense(32, activation='relu')(scalar_input)

# --- Merge All ---
merged = Concatenate()([x1, x2, x3])
bottleneck = Dense(64, activation='relu', name='embedding')(merged)

# --- Decoder (output = reconstruction of features) ---
x = Dense(256, activation='relu')(bottleneck)
x = Dense(512, activation='relu')(x)
decoded = Dense(X_features.shape[1], activation='sigmoid', name="output")(x)

# --- Full Model ---
model = Model(inputs=[feature_input, image_input, scalar_input], outputs=decoded)
model.compile(optimizer='adam', loss='mse')

model.summary()
```

Train The Model

```python
model.fit(
    x=[X_features, X_images, X_scalars],
    y=X_features,  # Autoencoder target = input features
    epochs=50,
    batch_size=16,
    validation_split=0.1
)

```

STEP 5: Save the Model & Extract Embeddings

```python
# Save full model
model.save("song_autoencoder.keras")

# Save embedding layer extractor
embedding_model = Model(
    inputs=model.inputs,
    outputs=model.get_layer("embedding").output
)
embedding_model.save("song_embedding_model.keras")

# Extract & save embeddings
embeddings = embedding_model.predict([X_features, X_images, X_scalars])
np.save("song_embeddings.npy", embeddings)
np.save("song_files.npy", X_files)
```

```python
from sklearn.cluster import KMeans
# Load your saved embeddings
embeddings = np.load('song_embeddings.npy')

# Create and train KMeans
kmeans = KMeans(n_clusters=10, random_state=42)
kmeans.fit(embeddings)

# Save the clustering model
joblib.dump(kmeans, 'kmeans_song_cluster.joblib')
print("KMeans clustering model saved!")
```
